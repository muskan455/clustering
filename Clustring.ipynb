**Q.2 How does K-Means clustering algorithm work?**

Ans = K-Means clustering is an iterative algorithm that aims to partition a dataset into K distinct, non-overlapping subgroups (clusters). Here's a breakdown of the steps:

1.  **Initialization:**
    *   Choose the number of clusters, K.
    *   Randomly select K data points from the dataset as the initial centroids (the center of each cluster).

2.  **Assignment:**
    *   Calculate the distance from each data point to each of the K centroids.
    *   Assign each data point to the cluster whose centroid is closest to it.

3.  **Update:**
    *   Recalculate the centroids of each cluster. The new centroid is the mean of all data points assigned to that cluster.

4.  **Iteration:**
    *   Repeat steps 2 and 3 until the centroids no longer move significantly or a maximum number of iterations is reached. This indicates that the clusters have stabilized.

The algorithm's goal is to minimize the within-cluster sum of squares (WCSS), which is the sum of the squared distances between each data point and its assigned centroid.

**Q.3 Explain the concept of a dendrogram in hierarchical clustering**

Ans= A dendrogram is a tree-like diagram that visualizes the steps and order in which data points or clusters are merged or split in hierarchical clustering.

Here's what you need to know about dendrograms:

*   **Branches and Leaves:** The "leaves" at the bottom of the dendrogram represent individual data points. As you move up the dendrogram, branches merge, indicating that the data points or clusters connected by that branch are being grouped together.
*   **Height of the Merge:** The vertical height of each merge point (the point where branches join) represents the distance or dissimilarity between the clusters being merged. The higher the merge point, the more dissimilar the clusters are.
*   **Cutting the Dendrogram:** To determine the final clusters, you can "cut" the dendrogram at a certain height. The number of vertical lines that intersect the horizontal cut line represents the number of clusters. Choosing the cut height allows you to control the granularity of the clusters.

Dendrograms are useful for:

*   Visualizing the hierarchical structure of the data.
*   Determining the optimal number of clusters by examining the distances at which clusters merge.
*   Understanding the relationships between individual data points and clusters.

**Q.4 What is the main difference between K-Means and Hierarchical Clustering?**

Ans = The main difference between K-Means and Hierarchical Clustering lies in their approach to forming clusters:

*   **K-Means Clustering:** This is a **partitioning** method. It requires you to specify the number of clusters (K) beforehand. The algorithm then iteratively assigns data points to one of the K clusters based on minimizing the distance to the cluster centroid. It produces a single set of K clusters.

*   **Hierarchical Clustering:** This is a **tree-based** method. It does not require you to specify the number of clusters in advance. Instead, it builds a hierarchy of clusters, either by starting with individual data points and merging them into larger clusters (agglomerative) or by starting with one large cluster and splitting it into smaller ones (divisive). The result is a dendrogram, which allows you to choose the number of clusters by cutting the tree at a desired level.

In summary, K-Means creates a flat partition of the data into a predefined number of clusters, while Hierarchical Clustering creates a nested sequence of clusters that can be visualized as a dendrogram, allowing for flexibility in choosing the number of clusters.

**Q.6 When would you use Silhouette Score in clustering?**

Ans = You would use the Silhouette Score in clustering to evaluate the quality of the clusters that have been created. It is a metric that measures how similar a data point is to its own cluster compared to other clusters. The Silhouette Score for a single data point ranges from -1 to +1:

*   A score close to +1 indicates that the data point is well-clustered and is far away from neighboring clusters.
*   A score close to 0 indicates that the data point is on or very close to the decision boundary between two neighboring clusters.
*   A score close to -1 indicates that the data point may have been assigned to the wrong cluster.

The average Silhouette Score across all data points in a dataset can be used to:

*   **Determine the optimal number of clusters:** You can run a clustering algorithm (like K-Means) for different values of K (number of clusters) and calculate the average Silhouette Score for each K. The value of K that yields the highest average Silhouette Score is often considered the optimal number of clusters.
*   **Compare different clustering algorithms:** You can compare the average Silhouette Scores obtained from different clustering algorithms on the same dataset to see which algorithm produces better-defined clusters.
*   **Assess the quality of individual clusters:** You can look at the Silhouette Scores of data points within each cluster to identify clusters that are not well-separated or contain potential outliers.

In summary, the Silhouette Score is a valuable tool for validating and comparing clustering results, helping you to understand how well-defined and separated your clusters are.

**Q.8 Why is feature scaling important in clustering algorithms like K-Means?**

Ans= Feature scaling is crucial for clustering algorithms like K-Means because they rely on distance calculations to group data points. Without scaling, features with larger values or wider ranges can disproportionately influence the distance metric, leading to biased clustering results.

Here's why it's important:

*   **Equal Contribution:** Scaling ensures that all features contribute equally to the distance calculations. If one feature has a range of 0-1000 and another has a range of 0-1, the feature with the larger range will dominate the distance calculation, regardless of its actual importance.
*   **Centroid Calculation:** In K-Means, centroids are calculated as the mean of the data points in a cluster. If features are not scaled, the centroid will be heavily influenced by features with larger values, potentially pulling the centroid towards points with extreme values in those features.
*   **Meaningful Distances:** Distance metrics like Euclidean distance are sensitive to the scale of the features. Scaling ensures that the distances between data points accurately reflect their similarity in the feature space, rather than being skewed by differences in scale.

Common scaling techniques include:

*   **Standardization (Z-score normalization):** Scales features to have a mean of 0 and a standard deviation of 1.
*   **Normalization (Min-Max scaling):** Scales features to a specific range, usually between 0 and 1.

By scaling features, you help clustering algorithms like K-Means to find more accurate and meaningful clusters based on the true underlying patterns in your data.

**Q.9 How does DBSCAN identify noise points?**

Ans = DBSCAN identifies noise points based on the concepts of density reachability and core points. Here's how it works:

1.  **Core Points:** A data point is considered a "core point" if there are at least a minimum number of other data points (MinPts) within a specified radius (epsilon, or ε) of it.
2.  **Density-Reachable:** A data point is "density-reachable" from a core point if it is within the epsilon radius of that core point or within the epsilon radius of another density-reachable point.
3.  **Density-Connected:** Two data points are "density-connected" if there is a core point from which both data points are density-reachable.
4.  **Clusters:** DBSCAN forms clusters by grouping together all density-connected data points.

**Noise Points:** Data points that are **not** density-reachable from any core point are considered noise points (or outliers). These are points that are in sparser regions of the dataset and do not have enough neighbors within the specified epsilon radius to be considered part of a dense cluster.

In essence, DBSCAN identifies noise points as those that are not part of any sufficiently dense region of the data.

**Q.10 Define inertia in the context of K-Means?**

Ans = In the context of K-Means clustering, **inertia** (also known as the within-cluster sum of squares or WCSS) is a metric used to evaluate how well the clusters are formed. It is calculated as the sum of the squared distances between each data point and the centroid of the cluster to which it is assigned.

Here's what inertia tells you:

*   **Lower Inertia is Better:** A lower inertia value generally indicates that the data points within each cluster are closer to their respective centroids, suggesting more compact and well-defined clusters.
*   **Measuring Cluster Tightness:** Inertia measures the "tightness" of the clusters. It reflects the variance within each cluster.

While inertia is a useful metric for evaluating K-Means clustering results, it's important to consider its limitations:

*   **Decreases with increasing K:** Inertia will always decrease as the number of clusters (K) increases, even if the additional clusters do not represent true structure in the data. This is because data points will naturally be closer to a centroid if there are more centroids.
*   **Doesn't account for cluster shape:** Inertia assumes spherical clusters and may not be the best metric for evaluating clusters of irregular shapes.

Therefore, inertia is often used in conjunction with other evaluation metrics, such as the Silhouette Score, and visual inspection of the clusters to determine the optimal number of clusters and assess the overall quality of the clustering.

**Q.11 What is the elbow method in K-Means clustering?**

**Q.11 What is the elbow method in K-Means clustering?**

Ans = The elbow method is a heuristic used in K-Means clustering to help determine the optimal number of clusters (K). It involves running the K-Means algorithm for a range of K values and calculating the inertia (within-cluster sum of squares) for each value of K.

Here's how the elbow method works:

1.  **Run K-Means for a range of K:** Perform K-Means clustering for a series of K values (e.g., from 1 to 10).
2.  **Calculate Inertia:** For each value of K, calculate the inertia, which is the sum of the squared distances of each data point to its assigned cluster centroid.
3.  **Plot the results:** Plot the inertia values against the corresponding K values.
4.  **Look for the "elbow":** The plot will typically show a curve that decreases rapidly at first and then starts to flatten out. The "elbow" point on the curve is the point where the rate of decrease sharply changes. This point is often considered to be the optimal number of clusters.

**Why the "elbow"?**

As you increase the number of clusters (K), the inertia will always decrease because the data points will be closer to their respective centroids. However, at some point, adding more clusters provides diminishing returns in terms of reducing inertia. The elbow point represents the trade-off between having too few clusters (high inertia) and having too many clusters (minimal gain in reducing inertia, potentially leading to overfitting).

**Limitations of the elbow method:**

*   **Subjectivity:** The "elbow" is not always clearly defined, and identifying it can be subjective.
*   **Doesn't work for all datasets:** Some datasets may not exhibit a clear elbow point.

Despite its limitations, the elbow method is a useful visual tool for getting an initial estimate of the optimal number of clusters in K-Means. It is often used in conjunction with other evaluation metrics and domain knowledge to make the final decision on the number of clusters.

**Q.12 Describe the concept of "density" in DBSCAN?**

Ans = In DBSCAN (Density-Based Spatial Clustering of Applications with Noise), "density" is a core concept used to define and discover clusters. Unlike partition-based methods like K-Means that rely on distances to centroids, DBSCAN groups together data points that are closely packed together in the data space.

The concept of density in DBSCAN is defined by two parameters:

1.  **Epsilon (ε):** This is a radius that defines a neighborhood around each data point.
2.  **MinPts:** This is the minimum number of data points required within the epsilon radius of a point for that point to be considered a "core point."

Based on these parameters, DBSCAN identifies three types of points:

*   **Core Point:** A data point is a core point if it has at least `MinPts` number of other data points (including itself) within its epsilon radius. These points are in the interior of a dense region.
*   **Border Point:** A data point is a border point if it has fewer than `MinPts` neighbors within its epsilon radius but is within the epsilon radius of a core point. These points are on the edge of a dense region.
*   **Noise Point:** A data point is a noise point if it is neither a core point nor a border point. These points are in sparse regions and are considered outliers.

DBSCAN then forms clusters by connecting core points that are within each other's epsilon radius and including the border points that are density-reachable from these core points. The concept of density allows DBSCAN to discover clusters of arbitrary shapes and to identify noise points effectively.

**Q.13 Can hierarchical clustering be used on categorical data?**

    *   **Simple Matching Coefficient:** Measures the proportion of attributes that are different between two data points.

*   **Converting to Numerical Representations:** In some cases, you might be able to convert categorical data into numerical representations based on domain knowledge or techniques like target encoding, although this should be done carefully to avoid introducing bias.

Once you have a way to calculate the dissimilarity or distance between data points with categorical features, you can apply hierarchical clustering as usual, using the dissimilarity matrix as input.

It's important to choose the appropriate preprocessing technique and dissimilarity measure based on the nature of your categorical data and the goals of your clustering analysis.

**Q.15 Explain the term "linkage criteria" in hierarchical clustering?**

Ans = In hierarchical clustering, "linkage criteria" (also known as the linkage method) determine how the distance or dissimilarity between two **clusters** is calculated when they are being considered for merging (in agglomerative clustering) or splitting (in divisive clustering). It's essentially the rule that governs how the algorithm decides which clusters to combine or divide at each step of the hierarchy.

Here are some common linkage criteria:

*   **Single Linkage (Minimum Linkage):** The distance between two clusters is defined as the **minimum** distance between any single data point in one cluster and any single data point in the other cluster. This method can sometimes create "chained" clusters.
*   **Complete Linkage (Maximum Linkage):** The distance between two clusters is defined as the **maximum** distance between any single data point in one cluster and any single data point in the other cluster. This method tends to produce more compact, spherical clusters.
*   **Average Linkage:** The distance between two clusters is defined as the **average** distance between all pairs of data points, where one point is from each cluster. This method is a compromise between single and complete linkage.
*   **Centroid Linkage:** The distance between two clusters is defined as the distance between the **centroids** (means) of the two clusters.
*   **Ward's Method:** This method aims to minimize the variance within each cluster. It calculates the distance between two clusters based on the increase in the total within-cluster sum of squares (inertia) that would result from merging them.

The choice of linkage criteria can significantly impact the shape and structure of the resulting dendrogram and, consequently, the clusters that are formed. The best linkage method to use depends on the nature of your data and the type of clusters you are trying to find.

**Q.16 Why might K-Means clustering perform poorly on data with varying cluster sizes or densities**

**Q.16 Why might K-Means clustering perform poorly on data with varying cluster sizes or densities**

Ans = K-Means clustering might perform poorly on data with varying cluster sizes or densities primarily because of its underlying assumptions and how it calculates cluster assignments and centroids:

*   **Assumption of Spherical and Equal-Sized Clusters:** K-Means inherently assumes that clusters are roughly spherical and of similar size. It works by minimizing the within-cluster sum of squares (inertia), which effectively tries to create compact, equally sized clusters around the centroids. When clusters have different shapes or significantly different numbers of data points, K-Means can struggle to accurately partition them. It might split large, dense clusters or merge smaller, sparser ones.

*   **Centroid as the Mean:** K-Means uses the mean of the data points in a cluster as its centroid. This works well for spherical clusters, but for clusters with irregular shapes or varying densities, the mean might not be a representative center. A centroid calculated from a dense region might be pulled towards a sparser region within the same cluster, or vice versa, leading to incorrect assignments.

*   **Distance-Based Assignment:** K-Means assigns each data point to the cluster whose centroid is closest. This distance-based approach doesn't consider the density of the surrounding data points. In data with varying densities, a point in a sparser but distinct cluster might be closer to the centroid of a denser, neighboring cluster and thus be incorrectly assigned.

*   **Sensitivity to Outliers:** While not directly related to varying sizes and densities, outliers can disproportionately influence the centroid calculation, especially in smaller or less dense clusters, further distorting the cluster boundaries.

In contrast, density-based algorithms like DBSCAN are better suited for data with varying cluster shapes and densities because they define clusters based on the density of data points rather than distances to centroids.

**Q.17  What are the core parameters in DBSCAN, and how do they influence clustering**

Ans = DBSCAN (Density-Based Spatial Clustering of Applications with Noise) has two core parameters that significantly influence the clustering results:

1.  **Epsilon (ε or `eps`):** This parameter defines the maximum distance between two data points for one to be considered as being in the neighborhood of the other. It essentially sets the radius of the neighborhood around each data point.

    *   **Influence:**
        *   A **larger** `eps` value will result in larger neighborhoods. This can lead to larger clusters being formed and potentially fewer noise points, as more points will fall within the neighborhood of core points. However, a very large `eps` can cause multiple distinct clusters to be merged into a single large cluster.
        *   A **smaller** `eps` value will result in smaller neighborhoods. This can lead to smaller clusters being formed and potentially more noise points, as fewer points will have enough neighbors to be considered core points. A very small `eps` might break down a single large cluster into multiple smaller ones or identify many points as noise.

2.  **MinPts:** This parameter defines the minimum number of data points (including the point itself) required within a point's epsilon neighborhood for that point to be considered a "core point."

    *   **Influence:**
        *   A **larger** `MinPts` value makes it harder for a point to be a core point. This means that only denser regions of the data will be considered as clusters, leading to fewer and potentially more distinct clusters. It also increases the number of points that are classified as noise.
        *   A **smaller** `MinPts` value makes it easier for a point to be a core point. This can lead to sparser regions being considered as clusters, potentially merging distinct clusters that are close to each other. It also decreases the number of points classified as noise. A `MinPts` of 1 effectively turns DBSCAN into hierarchical clustering with single linkage.

**Choosing the parameters:**

Selecting the appropriate values for `eps` and `MinPts` is crucial for obtaining meaningful clustering results with DBSCAN. This often involves:

*   **Domain Knowledge:** Understanding the data and the expected density of clusters can help in setting initial values.
*   **Visualization:** Visualizing the data can provide insights into appropriate values for `eps` and `MinPts`.
*   **Heuristics:** Techniques like k-distance plots (plotting the distance of each point to its k-th nearest neighbor) can help in determining a suitable value for `eps`.
*   **Experimentation:** Trying different combinations of `eps` and `MinPts` and evaluating the results using metrics or visual inspection.

In summary, `eps` controls the maximum distance for neighborhood inclusion, and `MinPts` controls the minimum density required to form a cluster. By adjusting these parameters, you can tune DBSCAN to find clusters of different densities and sizes in your data.

**Q.18 How does K-Means++ improve upon standard K-Means initialization**

**Q.18 How does K-Means++ improve upon standard K-Means initialization**

Ans = K-Means++ is an improved initialization technique for the K-Means clustering algorithm. Standard K-Means initialization often involves randomly selecting K data points as the initial centroids. While simple, this can sometimes lead to suboptimal clustering results, especially if the initial centroids are poorly chosen (e.g., all close together or in a sparse region).

K-Means++ addresses this issue by using a more intelligent approach to select the initial centroids. Here's how it works:

1.  **Select the first centroid:** Randomly select one data point from the dataset to be the first centroid.
2.  **Calculate distances:** For each remaining data point, calculate the distance to the nearest centroid that has already been selected.
3.  **Select the next centroid:** Select the next centroid from the remaining data points with a probability proportional to the square of its distance to the nearest existing centroid. This means that data points that are farther away from the already selected centroids have a higher probability of being chosen as the next centroid.
4.  **Repeat:** Repeat steps 2 and 3 until K centroids have been selected.

**How this improves standard K-Means:**

*   **Better spread of initial centroids:** By favoring data points that are farther away from existing centroids, K-Means++ ensures that the initial centroids are more spread out across the data space. This helps to avoid the problem of all centroids being initialized in a small region.
*   **Reduced chance of suboptimal solutions:** A better initial placement of centroids increases the likelihood that the K-Means algorithm will converge to a better (lower inertia) clustering solution. It reduces the chances of getting stuck in local optima.
*   **Faster convergence:** With more strategically placed initial centroids, the K-Means algorithm often converges faster.

In summary, K-Means++ is a smarter initialization method that selects initial centroids in a way that promotes a better spread and improves the chances of finding a good clustering solution compared to random initialization.

**Q.19  What is agglomerative clustering?**

Ans = Agglomerative clustering is a type of hierarchical clustering that follows a "bottom-up" approach. It starts with each data point as a separate cluster and then iteratively merges the closest pairs of clusters until only one cluster remains or a stopping criterion is met.

Here's a breakdown of the steps:

1.  **Initialization:** Each data point is treated as a single cluster.
2.  **Calculate Dissimilarity:** Calculate the dissimilarity (or distance) between all pairs of clusters. The choice of how to calculate the distance between clusters is determined by the "linkage criteria" (e.g., single linkage, complete linkage, average linkage, Ward's method).
3.  **Merge Clusters:** Merge the two closest (most similar) clusters based on the chosen linkage criteria.
4.  **Repeat:** Repeat steps 2 and 3 until all data points are in a single cluster or the desired number of clusters is reached.

The process of merging clusters can be visualized using a dendrogram, which shows the hierarchy of merges. By cutting the dendrogram at a certain height, you can obtain a specific number of clusters.

Agglomerative clustering is a popular method for hierarchical clustering because it is relatively easy to understand and implement. However, it can be computationally expensive for large datasets due to the need to calculate and update the dissimilarity matrix at each step.

**Q.20  What makes Silhouette Score a better metric than just inertia for model evaluation?**

Ans = While inertia is a useful metric in K-Means for evaluating the compactness of clusters, the Silhouette Score is often considered a better metric for overall model evaluation because it takes into account both the **cohesion** within clusters and the **separation** between clusters.

Here's why Silhouette Score provides a more comprehensive evaluation:

*   **Measures both Cohesion and Separation:**
    *   **Inertia** only measures the cohesion within clusters (how close points are to their own cluster's centroid). It doesn't consider how well separated the clusters are from each other.
    *   **Silhouette Score** measures both cohesion (how similar a data point is to its own cluster) and separation (how dissimilar it is to other clusters). It quantifies how well a data point fits into its assigned cluster and how poorly it fits into neighboring clusters.

*   **Handles Increasing K:**
    *   **Inertia** will always decrease as the number of clusters (K) increases, making it difficult to determine the optimal K based on inertia alone. You need to look for an "elbow" in the inertia plot, which can be subjective.
    *   **Silhouette Score** can help in determining the optimal number of clusters. A higher average Silhouette Score generally indicates a better clustering solution. You can calculate the average Silhouette Score for different values of K and choose the K that maximizes this score.

*   **Provides Insight into Individual Data Points:**
    *   **Inertia** gives a single value for the entire clustering result.
    *   **Silhouette Score** provides a score for each individual data point, which can help in identifying data points that are poorly clustered or are potentially outliers.

In essence, the Silhouette Score offers a more balanced perspective on the quality of clustering by considering both how tight the clusters are and how well-separated they are from each other. This makes it a more robust metric for evaluating and comparing different clustering solutions, especially when trying to determine the optimal number of clusters.

Generate synthetic data with 300 samples, 4 centers, a cluster standard deviation of 0.5, and a random state of 42 using `make_blobs`.



````python
from sklearn.datasets import make_blobs

# Generate synthetic data
X, y = make_blobs(n_samples=300, centers=4, cluster_std=0.5, random_state=42)
````

## Calculate silhouette scores

### Subtask:
Calculate the silhouette score for K-Means clustering with k from 2 to 5.


**Reasoning**:
Calculate the silhouette score for K-Means clustering with k from 2 to 5.



````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Initialize an empty list to store the silhouette scores
silhouette_scores = []

# Loop through values of k from 2 to 5 (inclusive)
for k in range(2, 6):
    # Instantiate a KMeans object
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)

    # Fit the KMeans model to the generated data
    kmeans.fit(X)

    # Predict the cluster labels for the data
    y_pred = kmeans.predict(X)

    # Calculate the silhouette score
    score = silhouette_score(X, y_pred)

    # Append the silhouette score to the list
    silhouette_scores.append(score)

print(silhouette_scores)
````

## Visualize silhouette scores

**Reasoning**:
Create a bar chart to display the silhouette scores for each value of k.

No need to specify centroids (unlike K-Means)

Used to reduce high-dimensional data into fewer components

Makes visualization possible and helps clustering perform better

**Q.44  Load the Iris dataset and plot the Silhouette Coefficient for each sample after KMeans clustering**

````python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples, silhouette_score

# Step 1: Load dataset
iris = load_iris()
X = iris.data

# Step 2: Apply KMeans clustering
kmeans = KMeans(n_clusters=3, random_state=42)
y_kmeans = kmeans.fit_predict(X)

# Step 3: Compute Silhouette Coefficients
sample_silhouette_values = silhouette_samples(X, y_kmeans)
avg_silhouette = silhouette_score(X, y_kmeans)

print(f"Average Silhouette Score: {avg_silhouette:.3f}")

# Step 4: Plot Silhouette Coefficients for each sample
plt.figure(figsize=(8,6))
y_lower = 10
for i in range(3):  # 3 clusters
    ith_cluster_silhouette_values = sample_silhouette_values[y_kmeans == i]
    ith_cluster_silhouette_values.sort()

    size_cluster_i = ith_cluster_silhouette_values.shape[0]
    y_upper = y_lower + size_cluster_i

    plt.fill_betweenx(np.arange(y_lower, y_upper),
                      0, ith_cluster_silhouette_values,
                      alpha=0.7)

    plt.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))
    y_lower = y_upper + 10

plt.axvline(x=avg_silhouette, color="red", linestyle="--", label=f"Average = {avg_silhouette:.2f}")
plt.title("Silhouette Plot for KMeans Clustering (Iris Dataset)")
plt.xlabel("Silhouette Coefficient Values")
plt.ylabel("Cluster Labels")
plt.legend()
plt.show()
````

| Step                                       | Description                                               |
| ------------------------------------------ | --------------------------------------------------------- |
| `silhouette_samples()`                     | Computes the silhouette coefficient for each sample       |
| `silhouette_score()`                       | Computes the **average silhouette score** for all samples |
| **Silhouette Coefficient Range:** -1 to +1 |                                                           |
| `fill_betweenx()`                          | Fills area plot for each cluster’s silhouette values      |


| Score Range       | Meaning                                       |
| ----------------- | --------------------------------------------- |
| **+1.0**          | Sample is well inside its cluster (very good) |
| **0.0**           | Sample is on or near the decision boundary    |
| **Negative (-1)** | Sample is likely in the wrong cluster         |

| --------------------------- | ------------------------------------------------------------------------- |
| `make_blobs()`              | Generates synthetic cluster data with 4 centers                           |

| `fit_predict()`             | Fits the model and returns predicted cluster labels                       |


| Linkage      | How It Works                              | Characteristics                    |
| ------------ | ----------------------------------------- | ---------------------------------- |
| **single**   | Minimum distance between points           | May form long “chain” clusters     |
| **complete** | Maximum distance between points           | Produces compact, tight clusters   |
| **average**  | Mean distance between all pairs of points | Balanced between single & complete |
| **ward**     | Minimizes variance between clusters       | Works best for spherical clusters  |

**Q.48  Load the Digits dataset, reduce dimensions using t-SNE, then apply Agglomerative Clustering and plot the
